## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

This is a ranking/scoring task. My goal is to produce a continuous score that orders all eligible pages from most to least worth a reviewer's limited attention. A reviewer works down a ranked list until their capacity runs out (e.g. top 50), so what matters is getting the order right at the top, not getting every page's label individually correct the way classification would.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

My proxy target is **is_declining_label**, defined as **trend_direction == "down"**.This comes from a defined rule, **trend_direction** is itself derived from **trend_pct** which is calculated from the current data window.I know this is a weaker "beginner proxy" for now, and a stronger version would use a real future-window outcome instead.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

My success metric is precision@50: of the top 50 pages my model ranks highest, how many are actually declining-with-demand pages? I'm choosing this over plain accuracy because this is a ranked-review problem — a reviewer only has capacity to check a limited number of pages, so what matters is whether the top of the list is trustworthy, not how the model performs across all 30,000 pages equally. The starter pipeline already gives me a number to beat: the rule-based baseline reaches 0.240 precision@50 (about 12 of the top 50 are real hits), so any model I build needs to clear that bar to prove it's adding value.
