## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Decision: Which content pages should a content/SEO reviewer prioritize for refresh, expansion, protection, or pruning, given limited review time?
Who acts, and how: A content strategist or SEO reviewer at a FlyRank client, who has capacity to manually review only a small number of pages (e.g. the top 20-50) each cycle. They use the ranked list to decide where to spend that time first, instead of reviewing pages in an arbitrary or alphabetical order.
Cost of a wrong call: If the model surfaces a page that isn't actually worth reviewing (a false positive), the reviewer wastes limited time on a low-value page while a genuinely declining or high-opportunity page goes unreviewed. If it misses a real opportunity (a false negative), a page that's actually losing visibility or leaving clicks on the table keeps quietly underperforming, unnoticed, until the next review cycle.

```python

import os, sys, subprocess, pandas as pd
IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"], check=True)
elif os.path.basename(os.getcwd()) == "notebooks":
    os.chdir("..")  # move from notebooks/ to the repo root

print("Working dir:", os.getcwd())
assert os.path.exists("data/raw/content_refresh_anonymized.csv"), "starter CSV not found — are you at the repo root?"
print("Starter data found. You're ready.")

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

df = df[(df["impressions_90d"] > 0) & (df["content_age_days"] >= 90)]
df = df.drop_duplicates(subset="content_id")

print("Total eligible content items:", len(df))
print("Pages currently declining (trend_direction == 'down'):",
      (df["trend_direction"] == "down").sum())
print("Pages that are both declining AND getting real demand "
      "(impressions_90d >= 100):",
      ((df["trend_direction"] == "down") & (df["impressions_90d"] >= 100)).sum())
```

Out of 30,000 eligible content items, 16,262 (about 54%) are currently flagged as declining. Of those, 13,152 (about 81% of the declining pages) have meaningful search demand (100+ impressions in the last 90 days) — meaning most of the decline isn't happening on dead, low-traffic pages, it's happening on pages people are still actually finding. That's a large, high-signal pool to build and validate a ranking system against, and it's exactly the kind of problem where prioritizing which page to review first has real value.

## 4. Careful words: what I can and can't claim

What I can claim: which pages observationally show signs of decline, thinness, or missed CTR opportunity relative to similar pages; a ranked, evidence-backed priority order for review; directional patterns in which signals associate with these opportunities.
What I cannot claim: that refreshing a flagged page will cause it to recover (that requires a controlled experiment, which this data doesn't provide); that any result reflects Google's actual ranking algorithm; that a client, page, or query in this dataset corresponds to any real, identifiable entity.
