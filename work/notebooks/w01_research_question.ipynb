## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Decision: Which content pages should a content/SEO reviewer prioritize for refresh, expansion, protection, or pruning, given limited review time?
Who acts, and how: A content strategist or SEO reviewer at a FlyRank client, who has capacity to manually review only a small number of pages (e.g. the top 20-50) each cycle. They use the ranked list to decide where to spend that time first, instead of reviewing pages in an arbitrary or alphabetical order.
Cost of a wrong call: If the model surfaces a page that isn't actually worth reviewing (a false positive), the reviewer wastes limited time on a low-value page while a genuinely declining or high-opportunity page goes unreviewed. If it misses a real opportunity (a false negative), a page that's actually losing visibility or leaving clicks on the table keeps quietly underperforming, unnoticed, until the next review cycle.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

df = df[(df["impressions_90d"] > 0) & (df["content_age_days"] >= 90)]
df = df.drop_duplicates(subset="content_id")

print("Total eligible content items:", len(df))
print("Pages currently declining (trend_direction == 'down'):",
      (df["trend_direction"] == "down").sum())
print("Pages that are both declining AND getting real demand "
      "(impressions_90d >= 100):",
      ((df["trend_direction"] == "down") & (df["impressions_90d"] >= 100)).sum())
```
